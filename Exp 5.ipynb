```python
description = input("Enter report description: ").lower()
audience = input("Enter intended audience: ").lower()

if any(word in description + audience for word in
       ["business risk", "executive", "management", "long-term", "strategy"]):
    print("Report Type: Strategic")

elif any(word in description + audience for word in
         ["attack method", "tactics", "techniques", "ioc", "indicators"]):
    print("Report Type: Tactical")

elif any(word in description + audience for word in
         ["incident", "active attack", "ip address", "urgent", "security analyst"]):
    print("Report Type: Operational")

else:
    print("Report Type: Unclassified")
```
