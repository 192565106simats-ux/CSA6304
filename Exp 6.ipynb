```python
from urllib.parse import urlparse
import ipaddress

url = input("Enter URL: ").strip()
parsed = urlparse(url if "://" in url else "http://" + url)
host = parsed.hostname or ""

score = 0
keywords = ["login", "verify", "update", "secure", "bank", "password"]

if any(word in url.lower() for word in keywords):
    print("Suspicious keyword detected")
    score += 1

try:
    ipaddress.ip_address(host)
    print("IP-based domain detected")
    score += 1
except ValueError:
    pass

if len(host.split(".")) > 3:
    print("Excessive subdomains detected")
    score += 1

if score >= 2:
    print("Result: Potentially Phishing URL")
else:
    print("Result: No major phishing indicators found")
```
