```python
import re
from collections import Counter

logs = [
    "INFO User login successful",
    "ERROR Database connection failed",
    "INFO User login successful",
    "WARNING High memory usage",
    "ERROR Database connection failed"
]

# Logstash: Parse raw logs
events = []
for log in logs:
    match = re.match(r"(\w+) (.*)", log)
    if match:
        events.append({
            "level": match.group(1),
            "message": match.group(2)
        })

# Elasticsearch: Index structured events
index = {i: event for i, event in enumerate(events)}

print("Indexed Events:")
for event in index.values():
    print(event)

# Kibana: Aggregate log levels
summary = Counter(event["level"] for event in index.values())

print("\nLog Summary:")
for level, count in summary.items():
    print(level, ":", count)
```
