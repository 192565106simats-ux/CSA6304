```python
network_policy = {
    "Low-Trust": ["Internet"],
    "Corporate": ["Internet", "Corporate"],
    "Sensitive": ["Sensitive"]
}

def check_access(source, destination):
    if destination in network_policy.get(source, []):
        print("Access ALLOWED:", source, "->", destination)
    else:
        print("Access BLOCKED:", source, "->", destination)

check_access("Low-Trust", "Internet")
check_access("Low-Trust", "Sensitive")
check_access("Corporate", "Sensitive")
check_access("Sensitive", "Sensitive")
```
